Day 2 EDA

```python
new_df = df.copy()
```

```python
new_df.isnull().sum()
```

```python

# filnna()
new_df["age"].median()
```

```python
new_df["age"].fillna(new_df["age"].median()).astype(int)
```

```python
new_df["gender"].mode()
```

```python
new_df["gender"]=new_df["gender"].fillna(new_df["gender"].mode()[0])
```

```python
new_df["attendance_percent"].median()
```

```python
new_df["attendance_percent"].fillna(new_df["attendance_percent"].median()).astype(int)
```

```python
new_df["exam_score"].median()
```

```python
new_df["exam_score"].fillna(new_df["exam_score"].median()).astype(int)
```

```python
new_df.duplicated().sum()
```

```python
new_df.head()
```

```python
new_df["gender"] = new_df["gender"].str.strip().str.lower().map({
    "male": "Male",
    "female": "Female"
})
```

```python
new_df["gender"].value_counts()
```

```python
new_df.head()
```

```python
new_df.loc[new_df["name"] == "Yusuf Yusuf", "gender"] = "Male"
```

```python
new_df[new_df["name"] == "Yusuf Yusuf"]
```

```python
new_df.loc[
    new_df["name"] == "Yusuf Hassan",
    "gender"
] = "Male"
```

```python
new_df[new_df["name"] == "Yusuf Hassan"]
```

```python
new_df["gender"].value_counts()
```

```python
new_df["school"] = new_df["school"].str.strip()
new_df["school"] = new_df["school"].str.lower()

```
